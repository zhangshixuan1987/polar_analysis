# SAM Index & Lead-Lag Workflow

This notebook drives the SAM (Southern Annular Mode) analysis workflow, which includes:
1. Generating SAM EOF-based or PSL-based indices from model outputs and observations.
2. Generating time series plots for the SAM indices.
3. Performing lead-lag regression and correlation analysis.

```python
import os
import sys
import glob
import collections

# Ensure our shared libraries are discoverable
sys.path.append("..")
from polar_utils.common import Case
from polar_utils.sam import (
    run_sam_eof_index_generation,
    run_sam_psl_index_generation,
    run_sam_leadlag_analysis
)
```

```python
# Global Configuration Paths
top_path = "/lcrc/group/e3sm2/ac.szhang/E3SMv21_testings/v3_polar_paper"
run_path = "/lcrc/group/e3sm/ac.szhang/acme_scratch/data/pcmdi/model/monthly"
run_mask = "/lcrc/group/e3sm/ac.szhang/acme_scratch/data/pcmdi/model/fixed/sftlf"
data_dir = "/lcrc/group/e3sm/public_html/diagnostic_output/ac.szhang/E3SMv2_1/pcmdi"
model_root = "/lcrc/group/e3sm2/ac.szhang/E3SMv21_testings"

sam_region = {'west': -180., 'east': 180., 'south': -90., 'north': -20.}
l_check_sam_region = False
```

## 1. Run SAM EOF-based Index Generation

```python
out_path = os.path.join(top_path, "fig_data", "sam_analysis", "raw_index")
fig_path = os.path.join(top_path, "polar_analysis", "4_sam_analysis", "sam_index_ts", "figure")

# 1.1 Historical E3SM run
case_dict = collections.OrderedDict()
eof_dir = os.path.join(data_dir, 'diagnostic_results/variability_modes', 'e3sm', 'historical', 'v20241111/SAM/NOAA-20C')
eof_file = sorted(glob.glob(os.path.join(eof_dir, 'SAM_PSL_EOF1_monthly_e3sm_*_historical_0701_0701_*_cbf.nc')))[0]
case_dict["v2_1-SORRM"] = Case(eof_file, "PSL", "blue", "v2_1-SORRM")
run_sam_eof_index_generation(fig_path, out_path, "e3sm", "historical", "0701", "sam_cbf", "195001-201412", 
                             case_dict, sam_region, l_check_sam_region)

# 1.2 Control E3SM run
case_dict = collections.OrderedDict()
eof_dir = os.path.join(data_dir, 'diagnostic_results/variability_modes', 'e3sm', 'piControl', 'v20241111/SAM/NOAA-20C')
eof_file = sorted(glob.glob(os.path.join(eof_dir, 'SAM_PSL_EOF1_monthly_e3sm_*_piControl_1950_1950_*_cbf.nc')))[0]
case_dict["v2_1-SORRM"] = Case(eof_file, "PSL", "blue", "v2_1-SORRM")
run_sam_eof_index_generation(fig_path, out_path, "e3sm", "piControl", "1950", "sam_cbf", "080101-100012", 
                             case_dict, sam_region, l_check_sam_region)
```

## 2. Run SAM PSL-based Index Generation

```python
# 2.1 Historical E3SM run (using the combined SORRM simulation)
case_dict = collections.OrderedDict()
sorrm_files = os.path.join(model_root, "v2_1.SORRM.histssp370_0701/post/atm/180x360_aave/ts/monthly/10yr/PSL_*.nc")
case_dict["v2_1-SORRM"] = Case(sorrm_files, "PSL", "blue", "v2_1-SORRM")
run_sam_psl_index_generation(fig_path, out_path, "e3sm", "historical", "0701", "sam_gongwang", "195101-201412", 
                             case_dict, sam_region, l_check_sam_region)

# 2.2 Control E3SM run
case_dict = collections.OrderedDict()
control_files = os.path.join(model_root, "v2_1.SORRM.control/post/atm/180x360_aave/ts/monthly/50yr/PSL_*.nc")
case_dict["v2_1-SORRM"] = Case(control_files, "PSL", "blue", "v2_1-SORRM")
run_sam_psl_index_generation(fig_path, out_path, "e3sm", "piControl", "1950", "sam_gongwang", "080101-100012", 
                             case_dict, sam_region, l_check_sam_region)
```

```python
lead_lag_out = os.path.join(top_path, "fig_data", "sam_analysis", "lead_lag")
lead_lag_fig = os.path.join(top_path, "polar_analysis", "4_sam_analysis", "lead_lag_analysis", "figure")

case_dict = collections.OrderedDict()
case_dict["v2_1-SORRM"] = Case(sorrm_files, "PSL", "blue", "v2_1-SORRM")
run_sam_leadlag_analysis(lead_lag_fig, lead_lag_out, "e3sm", "historical", "0701", "sam_cbf", "195101-201412", 
                         case_dict, sam_region, "samSD_idx", l_check_sam_region)
```
